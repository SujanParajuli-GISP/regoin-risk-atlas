## Prepare geojson for GEE

```python
import geopandas as gpd
from pathlib import Path

# blocks = gpd.read_file("../data/boundaries/blocks_saptari.geojson")
blocks = gpd.read_file("../data/boundaries/districts_nepal.geojson")

# Keep only essential fields
blocks = blocks.reset_index(drop=True)
blocks["block_id"] = blocks.index.astype(int)
# blocks = blocks[["block_id", "OBJECTID","GaPa_NaPa", "geometry"]]
blocks = blocks[["block_id","fid_1","FIRST_DIST","geometry"]]

# Save cleaned file for GEE
OUT = Path("../data/boundaries")
# blocks.to_file(OUT / "blocks_saptari_gee.geojson", driver="GeoJSON")
blocks.to_file(OUT / "districts_nepal_gee.geojson", driver="GeoJSON")

# print("Prepared districts_nepal_gee.geojson for GEE")
print("Prepared districts_nepal_gee.geojson for GEE")
```
